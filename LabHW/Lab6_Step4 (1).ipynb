```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display

# --------------------------------
# Data path
# --------------------------------
basepath = Path.home() / "Downloads" / "datafolder"

indoorpath = basepath / "indoor.csv.csv"
outdoorpath = basepath / "outdoor1.csv.csv"
aq_inside_path = basepath / "aq_temp_inside.csv.csv"
aq_outside_path = basepath / "aq_temp_outside.csv.csv"

# --------------------------------
# Load data
# --------------------------------
indoor = pd.read_csv(indoorpath)           # Our Group Indoor
rundata = pd.read_csv(outdoorpath)         # Our Group Outdoor
aq_inside = pd.read_csv(aq_inside_path)    # Other Group Indoor
aq_inside = aq_inside.iloc[1:].reset_index(drop=True)
aq_outside = pd.read_csv(aq_outside_path)  # Other Group Outdoor
```

```python
plt.figure()

plt.scatter(indoor["Temp"], indoor["Humidity"], label="Our Indoor")
plt.scatter(aq_inside["Temperature"], aq_inside["Humidity"], label="Other Indoor")

plt.xlabel("Temperature")
plt.ylabel("Humidity")
plt.title("Temperature vs Humidity")

plt.legend()
plt.grid(True)

plt.show()
```

```python
#Hidden variable? --> TIME! Time needs to be equal at each measurement to properly do comparison.#
```

```python
#This matters because if we try to graph two sets of data and their times aren't equal at each point we're comparing, the difference could be caused by time and not actually the variables we're comparing.
```

```python
#This plot shows a negative correlation between temperature and humidity, and the relationships both are linear (both follow a straight line).  
```

```python
#Temp v. Pressure 
plt.figure()

plt.scatter(indoor["Temp"], indoor["Pressure"], label="Our Indoor")
plt.scatter(aq_inside["Temperature"], aq_inside["Pressure"], label="Other Indoor")

plt.xlabel("Temperature")
plt.ylabel("Pressure")
plt.title("Temperature vs Pressure")

plt.legend()
plt.grid(True)

plt.show()
```

```python
#No correlation. Since the slope of both the data sets is zero, there is no correlation.
```

```python
#Temp v. Gas (only for our group's dataset, other group didn't have this column in their set) 
plt.figure()

plt.scatter(indoor["Temp"], indoor["Gas"], label="Our Indoor")

plt.xlabel("Temperature")
plt.ylabel("Gas")
plt.title("Temperature vs Gas")

plt.legend()
plt.grid(True)

plt.show()
```

```python
#There is a positive correlation between Temperature and gas. This correlation is positively correlated at a greater amount than the negative correlation of temperature v. humidity.
```
